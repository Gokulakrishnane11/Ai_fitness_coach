# 🔥 Smart Fat Loss AI Coach — EDA & Analysis
Exploratory analysis of the synthetic fat-loss dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('dark_background')
df = pd.read_csv('../fat_loss_dataset.csv')
print(df.shape)
df.head()
```

```python
# Basic statistics
df.describe().round(2)
```

```python
# Target distribution
fig, ax = plt.subplots(figsize=(9, 5))
ax.hist(df['weekly_fat_loss_kg'], bins=50, color='#FF6B35', alpha=0.85, edgecolor='none')
ax.set_xlabel('Weekly Fat Loss (kg)')
ax.set_ylabel('Count')
ax.set_title('Distribution of Weekly Fat Loss')
plt.tight_layout()
plt.show()
```

```python
# Gender breakdown
df.groupby('gender')['weekly_fat_loss_kg'].describe().round(3)
```

```python
# Workout days vs fat loss
fig, ax = plt.subplots(figsize=(10, 5))
df.boxplot(column='weekly_fat_loss_kg', by='workout_days_per_week', ax=ax)
ax.set_title('Workout Days per Week vs Weekly Fat Loss')
ax.set_xlabel('Workout Days per Week')
ax.set_ylabel('Weekly Fat Loss (kg)')
plt.suptitle('')
plt.tight_layout()
plt.show()
```

```python
# Correlation matrix
numeric_cols = ['age','height_cm','weight_kg','workout_days_per_week',
                'daily_calories','protein_intake_g','daily_steps','tdee','weekly_fat_loss_kg']
corr = df[numeric_cols].corr()

fig, ax = plt.subplots(figsize=(11, 8))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, cmap='RdYlGn', center=0,
            annot=True, fmt='.2f', linewidths=0.5, ax=ax)
ax.set_title('Correlation Heatmap')
plt.tight_layout()
plt.show()
```

```python
# Top correlations with target
print('Correlations with weekly_fat_loss_kg:')
print(corr['weekly_fat_loss_kg'].sort_values(ascending=False).to_string())
```
